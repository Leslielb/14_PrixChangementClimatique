#### Budget des communes


```python
# Récupération de la table
table_name = "budget_per_compte_communes"
compte_table = con.sql(f"SELECT * FROM {table_name}")

budget = compte_table.df()
budget
```

#### SIREN Insee


```python
insee_siren = pd.read_csv(
    "C:/Users/lesli/kDrive2/Data/Data For Good/14_prixchangementclimatique/Sources/banatic-sireninsee-last.csv",
    sep=";",
    encoding="latin-1",
)
insee_siren.dtypes
```

```python
insee_siren["siren"] = insee_siren["siren"].astype("str")
insee_siren["len"] = insee_siren["siren"].str.len()
insee_siren.groupby(["len"])["insee"].nunique()
```

```python
insee_siren = insee_siren[["siren", "insee", "nom_com"]].copy()
insee_siren.drop_duplicates(subset="insee", inplace=True)
insee_siren
```

### Croisement fichier budget et INSEE pour filtrer sur les communes


```python
budget["siren"] = budget["siret"].astype("str").str[:9]
```

```python
# Fusion avec le référentiel des communes
budget = pd.merge(
    budget, insee_siren[["siren", "insee", "nom_com"]], on="siren", how="left"
)
```

```python
# Suppression des lignes sans nom de commune
budget = budget.loc[budget["nom_com"] != ""].drop(columns=("code_geo"))
```

### Cleaning


#### Ajout d'une colone année + code geo


```python
budget["annee_insee"] = budget["annee"].astype(str) + budget["insee"]
budget
```

```python
# Ajout d'une colonne prime basé sur les lignes "primes d asssurances" dans la colonne tupe de compte
budget["prime"] = np.where(
    budget["type_compte"] == "primes d assurances", budget["solde_debiteur"], 0
)
```

### Exploration


#### Groupement sur le budget total


```python
# Création d'un dataframe à partir du groupement des lignes
budget_tot = budget.groupby(
    ["annee_insee", "annee", "insee", "nom_com"],
    as_index=False,
).agg(budget=("solde_debiteur", "sum"), prime=("prime", "sum"))
budget_tot
```

```python
# Renommage de la colonne solde debiteur
budget_tot.rename(columns={"solde_debiteur": "budget"}, inplace=True)
```

```python
budget_tot.groupby("annee")["insee"].nunique()
```

#### Suppression des lignes avec 0 budget


```python
budget_tot = budget_tot.loc[budget_tot["budget"] > 0].copy()
```

```python
# Vérification qu'il n'y ait pas de doublons
duplicates = budget_tot[budget_tot.duplicated(subset=["annee_insee"], keep=False)]
duplicates
```

#### Calcul de la part de la prime dans le budget total


```python
# Remplacement des valeurs NaN en 0
budget_tot["prime"] = np.nan_to_num(budget_tot["prime"])
```

```python
# Vérification du nombre de lignes sans prime
budget_tot.loc[budget_tot["prime"] == 0]
```

```python
# Calcul de la part de la prime dans le budget total
budget_tot["part_prime"] = budget_tot["prime"] / budget_tot["budget"]
budget_tot
```

### Evolution du montant des primes de 2020 à 2024


#### Filtre des données


```python
# Filtre sur les lignes de 2020 à 2024
budget_20_24 = budget_tot.loc[
    (budget_tot["annee"] == 2020)
    | (budget_tot["annee"] == 2021)
    | (budget_tot["annee"] == 2022)
    | (budget_tot["annee"] == 2023)
    | (budget_tot["annee"] == 2024)
]
```

```python
# Transformation en pivot table
budget_pivot = budget_20_24.pivot_table(
    values=["budget", "prime", "part_prime"],
    index=["insee", "nom_com"],
    columns="annee",
    aggfunc="first",
)
```

```python
# Tranformation du pivot table en dataframe
budget_pivot = pd.DataFrame(budget_pivot.to_records())

# Renommage des colonnes
budget_pivot.rename(
    columns={
        "('budget', 2020)": "budget_20",
        "('budget', 2021)": "budget_21",
        "('budget', 2022)": "budget_22",
        "('budget', 2023)": "budget_23",
        "('budget', 2024)": "budget_24",
        "('prime', 2020)": "prime_20",
        "('prime', 2021)": "prime_21",
        "('prime', 2022)": "prime_22",
        "('prime', 2023)": "prime_23",
        "('prime', 2024)": "prime_24",
        "('part_prime', 2020)": "part_prime_20",
        "('part_prime', 2021)": "part_prime_21",
        "('part_prime', 2022)": "part_prime_22",
        "('part_prime', 2023)": "part_prime_23",
        "('part_prime', 2024)": "part_prime_24",
    },
    inplace=True,
)

budget_pivot
```

#### Calcul de l'évolution


```python
# Création d'une colonne qui vérifie que les colonnes montant de la prime de 2020 et 2024 sont > 0
budget_pivot["check_20_24"] = np.where(
    (budget_pivot["part_prime_20"] > 0) & (budget_pivot["part_prime_24"] > 0),
    True,
    False,
)
```

```python
# Filtre sur les lignes qui ont un montant de prime entre 2020 et 2024
budget_pivot_filter = (
    budget_pivot.loc[budget_pivot["check_20_24"] == True]
    .drop(columns=["check_20_24"])
    .copy()
)
```

```python
# Calcul de l'évolution entre 2020 (ou 2021) et 2024
budget_pivot_filter["evol_20_24"] = np.where(
    budget_pivot_filter["part_prime_20"] > 0.0001,
    (budget_pivot_filter["prime_24"] - budget_pivot_filter["prime_20"])
    / budget_pivot_filter["prime_20"],
    (budget_pivot_filter["prime_24"] - budget_pivot_filter["prime_21"])
    / budget_pivot_filter["prime_21"],
)

# Trie par ordre descendant de l'évolution
budget_pivot_filter.loc[budget_pivot_filter["evol_20_24"] <= 1].sort_values(
    ["evol_20_24"], ascending=False
).head(1000)
```

```python
# Export d'un fichier Excel avec les 1000 communes ayant l'évolution du montant de la prime la plus forte (capé à 100%)
budget_pivot_filter.loc[budget_pivot_filter["evol_20_24"] <= 1].sort_values(
    ["evol_20_24"], ascending=False
).head(1000).to_excel(
    "C:/Users/lesli/kDrive2/Data/Data For Good/14_prixchangementclimatique/Résultats/Top 1000 hausse de prime 2020-2024.xlsx"
)
```

#### Visualisation des résultats


```python
# Catégorisation du montant de budget de 2024
conditions_budget = [
    (budget_pivot_filter["budget_24"] < 100000),
    (budget_pivot_filter["budget_24"] < 500000),
    (budget_pivot_filter["budget_24"] < 1000000),
    (budget_pivot_filter["budget_24"] >= 1000000),
]

# Catégories
results_budget = [
    "1. < 100K",
    "2. 100K-500K",
    "3. 500K-1M",
    "4.> 1M",
]

# Création d'une colonne pour budget_cat à partir des conditions
budget_pivot_filter["budget_cat"] = np.select(conditions_budget, results_budget)

# Regroupement par catégories
budget_pivot_filter.groupby(["budget_cat"])["insee"].nunique()
```

```python
sort = sorted(budget_pivot_filter["budget_cat"].unique())

fig_dis = px.histogram(
    budget_pivot_filter.loc[budget_pivot_filter["evol_20_24"] <= 1],
    x="evol_20_24",
    color="budget_cat",
    category_orders={"budget_cat": sort},
)

fig_dis.update_layout(
    title="Distribution de l'évolution de la prime entre 2020 et 2024",
    yaxis_title="Nombre de communes",
    xaxis_title="Evolution entre 2020 et 2024 (%)",
    legend_title="Montant du budget",
)

fig_dis.show()
```
